```python
import matplotlib.pyplot as sss
```

```python
sss.scatter(x,y,color="red",label="data points")
sss.legend()
sss.grid(True)
sss.xlabel("language")
sss.ylabel("difficulty")
sss.show()
```

```python
sss.bar(x,y,color="red",label="data points")
sss.legend()
sss.grid(True)
sss.xlabel("language")
sss.ylabel("difficulty")
sss.show()
```

```python
sss.plot(x, y, color="red", marker="o", label="data points")
sss.xlabel("Language")
sss.ylabel("Difficulty")
sss.legend()
sss.grid(True)
sss.show()
```

```python
sss.boxplot(y)
sss.xlabel("Difficulty")
sss.ylabel("Values")
sss.title("Difficulty Box Plot")
sss.show()
```

```python
sss.hist(y, bins=5)
sss.xlabel("Difficulty")
sss.ylabel("Frequency")
sss.title("Difficulty Distribution")
sss.show()
```
