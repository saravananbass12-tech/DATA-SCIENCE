```python
import numpy as np
import pandas as pd
c=pd.read_csv("Teen_Mental_Health_Dataset.csv")
c
```

```python
skew_value =c["stress_level"].skew()
print("\nskewness value :",round(skew_value,4))
```

```python
if skew_value>0:
  print("positive skewness")
elif skew_value<0:
  print("negative skewness")
else:
  print("symmetrical")
```

```python
import matplotlib.pyplot as plt
from scipy.stats import binom
import numpy as np

n = 12
p = 0.6
k = 10
ans = binom.pmf(k, n, p)
print(ans)
```

```python
x = np.arange(0, n + 1)
pmf_values = binom.pmf(x, n, p)

plt.bar(x, pmf_values, color="lightblue", edgecolor="black", label="Binomial PMF")
prob = binom.pmf(k, n, p)
plt.bar(k, prob, color="orange", edgecolor="black", label=f"k = {k}")
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```
