```python
import pandas as pd
```

```python
x=pd.read_csv("programming_languages.csv")
x
```

```python
a=x.loc[1:3,"language":"popularity"]
a
```

```python
import matplotlib.pyplot as plt
plt.bar(x["language"],x["popularity"])
plt.xlabel("programming language")
plt.ylabel("popularity")
plt.title("languages")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.bar(x["language"],x["difficulty"])
plt.xlabel("programming language")
plt.ylabel("difficulty")
plt.title("languages")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.bar(x["language"],x["year_created"])
plt.xlabel("language")
plt.ylabel("year_created")
plt.title("languages")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.pie(
    x["popularity"],
    labels=x["language"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)
plt.title("Programming Language")
plt.show()
```
