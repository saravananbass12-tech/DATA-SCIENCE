```python
import numpy as np
from scipy import stats
data =[23,33,22,29,21,25,25,28,35,32]
ci= stats.t.interval(confidence=0.95,df=len(data)-1,loc=np.mean(data),scale=stats.sem(data))
print("95% cofidence interval")
print("lower:",ci[0])
print("upper:",ci[1])
```
